```python
from pyspark.sql.types import DoubleType

# Only run this if 'overall' is not already a numeric type
gold_df = gold_df.withColumn("overall", col("overall").cast(DoubleType()))
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Convert a small sample to Pandas for plotting (plotting millions of rows directly will crash the driver)
pdf_subset = gold_df.select("overall").sample(False, 0.1, seed=42).toPandas()

plt.figure(figsize=(8, 5))
sns.countplot(data=pdf_subset, x='overall', palette='viridis')
plt.title('Distribution of Ratings (Gold Dataset)')
plt.xlabel('Rating (1-5)')
plt.ylabel('Count')
plt.show()
```

```python
from pyspark.sql.functions import length

# Add a temporary column for review length
df_with_len = gold_df.withColumn("text_length", length(col("reviewText")))

# Sample and convert to Pandas
pdf_len = df_with_len.select("text_length").sample(False, 0.05, seed=42).toPandas()

plt.figure(figsize=(10, 6))
sns.histplot(pdf_len['text_length'], bins=50, kde=True, color='blue')
plt.title('Distribution of Review Text Length')
plt.xlabel('Number of Characters')
plt.ylabel('Frequency')
# Optional: zoom in on the bulk of the data (e.g., 0 to 2000 chars)
plt.xlim(0, 2000) 
plt.show()
```

```python
# Define sample size
sample_size = 300000

# To ensure your sample is 'drift-resistant', we sort by a column like 'reviewTime' 
# or use a stratified sample to get a mix of all rating types.
# Here is a simple random sample:
total_rows = gold_df.count()
fraction = sample_size / total_rows

df_sampled = gold_df.sample(withReplacement=False, fraction=fraction, seed=42).limit(sample_size)

print(f"Original Count: {total_rows}")
print(f"Sampled Count: {df_sampled.count()}")

# Quick check: does the sample have the same rating distribution?
display(df_sampled.groupBy("overall").count().orderBy("overall"))
```
