```python
from pyspark.sql.functions import col, count, when, expr

validation_results = gold_df.select([
    count(when(col(c).isNull() | (col(c) == ""), c)).alias(c) 
    if c != "overall" else 
    count(when(expr("try_cast(overall as double)").isNull(), "overall")).alias("overall")
    for c in ["reviewText", "overall", "asin", "reviewerID"]
])

display(validation_results)
```

```python
from pyspark.sql.functions import expr

# 1. Use try_cast to turn empty strings into NULL
# 2. Filter out those NULLs
# 3. Group and count
ratings_dist = gold_df.withColumn("overall_num", expr("try_cast(overall as double)")) \
                      .filter(col("overall_num").isNotNull()) \
                      .groupBy("overall_num").count().orderBy("overall_num")

display(ratings_dist)
```

```python
from pyspark.sql.functions import length

# Calculate character length for valid reviews only
review_lengths = gold_df.filter(col("reviewText").isNotNull() & (col("reviewText") != "")) \
                         .select(length(col("reviewText")).alias("char_length"))

display(review_lengths)
```
